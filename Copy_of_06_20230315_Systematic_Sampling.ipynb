# **Rata-rata Kayu per Are**

- Forester tertarik untuk mengestimasi rata-rata kayu per are untuk 520 luasan per are **(N = 520).**
- Dilakukan **1-in-25** systematic sampling.
- Didapatkan data sebagai berikut.

image.png

# 1. Estimasi rata-rata volume kayu per luasannya.

```python
# Load library
import numpy as np
import pandas as pd
from scipy import stats

import matplotlib.pyplot as plt
```

```python
# Ukuran Populasi (N)
n_pop = 520

# interval
k = 25
```

Input data :

```python
# masukkan data
index = np.array([4,  29,  54,  79, 104, 129, 154, 179, 204, 
                  229, 254, 279, 304, 329, 354, 379, 404, 429, 
                  454, 479, 504])

# volume
data = np.array([7030, 6720, 6850, 7210, 7150, 7370, 7000, 6930, 
                 6570, 6910, 7380, 7540, 6720, 6900, 7200, 7100, 
                 6860, 6800, 7050, 7420, 7090])
```

Cek korelasi antar data :

```python
fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(10, 7))

ax.scatter(index, data, marker="x", color="red")

# x = indekx; y = volume
ax.set_xlabel("Index")
ax.set_ylabel("Volume (in board feet)")

plt.show()
```

Karena tidak terdapat korelasi, artinya data yang kita miliki **random**. Oleh karena itu, mean dapat diestimasi seperti halnya pada **SRS**.

**Estimating Population Mean**

$$
\hat{\mu} = \cfrac{1}{n} \sum_{i=1}^{n} y_{i}
$$


```python
# define sample size
n = len(data)

# calculate mean
mu_hat = np.mean(data)
print(f"mu_hat = {mu_hat}")
```

# 2. Estimasi bound of error dari estimatornya (CI = 95%)

- Dapat kita cari dengan

$$
B = z_{\alpha/2} \sqrt{\widehat{\text{var}}(\hat{\mu}_{sys})}
$$

- Dengan

$$
\widehat{\text{var}}(\hat{\mu}_{sys}) = \left( 1 - \cfrac{n}{N} \right) \cfrac{s^{2}}{n}
$$

```python
# tentukan varians sampel
s_square = np.var(data, ddof = n-1)

# varians dari taksiran rata-rata populasi
var_hat = (1-(n/n_pop))*(s_square/n)

# bound of error untuk CI 95%
alpha = 0.05
z = stats.norm.ppf(1-(alpha/2))
b = z * np.sqrt(var_hat)

print(f"Bound of error = {b}")
```

```python
ub = mu_hat + b
lb = mu_hat - b

print(f"ub = {ub}")
print(f"lb = {lb}")
```
