```python
# Load data manipulation package
import numpy as np
import pandas as pd

# Load statistic package
from scipy import stats
```

# Case : Populasi Salmon

**Deskripsi**
- Dalam perikanan diperlukan manajemen yang baik untuk mengetahui kualitas dan kuantitas hasil.
- Salah satunya adalah dengan mengetahui jumlah ikan yang akan dihasilkan.
- Di suatu area, 10.000 ikan salmon tertangkap. Selanjutnya salmon-salmon tersebut diberi penanda kawat halus dan dilepaskan.
- Beberapa hari kemudian, tim riset kembali menangkap salmon di lahan yang sama.
- Hasilnya tertangkap 6000 salmon.
- Dengan alat *metal detector* terdapat 400 salmon yang memiliki kawat penanda.

## **1. Estimate the number of salmon in population with its bound on error**

```python
# Salmon yang tertangkap di awal dan diberi tanda
x_1 = 10_000 # sample pertama # X

# Salmon yang ditangkap kembali
y = 6_000 # sample kedua

# Salmon yang ditangkap kembali dalam kondisi ditandai
x_2 = 400 # tertanda # x
```

- Estimasi populasi salmon

$$
\hat{\tau} = \cfrac{y}{x} X
$$

```python
# total population
tau_hat = (y/x_2) * x_1

print(f"Estimasi populasi salmon pada lahan adalah : {tau_hat} salmon")
```

- Selanjutnya, menentukan **bound of error** dengan ${\alpha}=0.05$.

$$
d = t_{\alpha/2} \sqrt{\widehat{\text{var}}(\hat{\tau})}
$$

- dengan varians dari total populasi

$$
\widehat{\text{var}}(\hat{\tau}) = 
\cfrac{Xy(X-x)(y-x)}
{x^{3}}
$$

```python
# estimasi varians total populasi
a = x_1 * y * (x_1-x_2)
b = y - x_2
c = x_2**3

var_tau_hat = (a*b)/c

print(f"Estimasi varians dari total populasi ikan salmon : {var_tau_hat}")
```

```python
# menghitung z_stat @ 95% CI
alpha = 0.05
t_stat = stats.norm.ppf(1 - alpha/2.0)
print(f"t_stat                                               : {t_stat:.2f}")

# calculate the margin of error (d)
d_tau = t_stat* np.sqrt(var_tau_hat)
print(f"margin of error dari estimasi total populasi (d_tau) : {d_tau:.0f} salmon")
```

```python
l_b = tau_hat - d_tau
u_b = tau_hat + d_tau

print(l_b)
print(u_b)
```
