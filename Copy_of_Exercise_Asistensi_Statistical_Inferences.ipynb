# Koinku Seimbang atau Tidak Ya ?

Aku dan Adikku sedang bermain di depan rumah. Dalam 100 kali pelemparan, diperoleh kepala sebanyak 67 kali. Apakah koin yang aku dan adikku mainkan seimbang ? 

Pembahasan:  
1.  Definisikan terlebih dahulu $H_0$ dan $H_1$ terlebih dahulu
$$
H_0 : p_0 = 0.5 \\
H_1 : p_0 \neq 0.5
$$
2. Hitung $z_{\text{hitung}}$:
$$
z_{\text{hitung}} = \frac{\hat{p}-p_0}{\sqrt{p_0(1-p_0)/n}}
$$
Diperoleh dari soal $\hat{p} = \frac{67}{100}$ dan $n = 100$. Sehingga diperoleh $z_{\text{hitung}}$ sebagai berikut: 
$$
z_{\text{hitung}} = \frac{0.67-0.5}{\sqrt{0.5(1-0.5)/100}} = 3.4
$$
3. Hitung titik kritis  
Perhatikan karna $H_1 : p_0 \neq 0.5$ maka diperoleh bahwa uji ini merupakan uji *two-tailed*. Pada soal, tidak disampaikan secara eksplisit batas signifikansi, $\alpha$, yang akan digunakan. Maka diasumsikan $\alpha = 5\%$. Maka diperoleh titik kritisnya adalah 
$$
z_{0.025} = 1.96
$$
Perhatikan bahwa nilai $z_{\text{hitung}} > z_{0.025}$ sehingga, $H_0$ **ditolak** atau dengan kata lain, koin yang dimainkan tidak seimbang

Berikutnya dapat dihitung selang kepercayaan dari $\hat{p}$ sebagai berikut $$\hat{p}\pm z_{0.025}\sqrt{p_0(1-p_0)/n} \\
= 0.67 \pm 1.96\sqrt{0.5(1-0.5)/100} \\
= 0.67 \pm 0.098 \\
= (0.572, 0.768)$$

Tapi kelihatannya soal yang dikerjakan kurang menarik :( Mari kita lihat soal berikutnya  

```python
round(norm.ppf(0.025),2)
```

# Balada COD di Negara XYZ
Di suatu negara, COD atau *cash on delivery* sering menjadi permasalahan karena satu dan lain hal. Anda direkrut oleh suatu perusahaan yang menjual barang secara online melalui suatu *platform online*. Anda mendapat data sampel hasil penjualan sebanyak 300,000 barang terjual selama satu bulan. Barang yang dijual dengan sistem COD sebanyak 125,000 dengan sebanyak 2,525 barang dikembalikan karena alasan lain. Anda ingin mengambil suatu kesimpulan apakah persentase dari penjualan dengan sistem COD berhasil lebih kecil dari 99%. Apakah anda dapat menarik kesimpulan tersebut ? 

Pembahasan:  
1.  Definisikan terlebih dahulu $H_0$ dan $H_1$ terlebih dahulu
$$
H_0 : p_0 \leq 0.99 \\
H_1 : p_0 > 0.99
$$
2. Hitung $z_{\text{hitung}}$:  
$$
z_{\text{hitung}} = \frac{\hat{p}-p_0}{\sqrt{p_0(1-p_0)/n}}
$$
Dari soal diperoleh bahwa nilai $\hat{p} = \frac{125,000-2,525}{125,000} = 0.9798$ Sehingga diperoleh $z_{\text{hitung}}$ sebagai berikut
$$
z_{\text{hitung}} = \frac{0.9798-0.99}{\sqrt{0.99(1-0.99)/125,000}} \approx -36.2441
$$
3. Hitung titik kritis  
Perhatikan karna $H_1 : p_0 > 0.99$ maka diperoleh bahwa uji ini merupakan uji *one-tailed*. Pada soal, tidak disampaikan secara eksplisit batas signifikansi, $\alpha$, yang akan digunakan. Maka diasumsikan $\alpha = 5\%$. Maka diperoleh titik kritisnya adalah 
$$
z_{0.05} = 1.64
$$
Perhatikan bahwa nilai $z_{\text{hitung}} < z_{0.05}$ sehingga, $H_0$ **tidak ditolak** atau dengan kata lain, bahwa penjualan dengan sistem COD berhasil lebih kecil dari 99%

Berikutnya dapat dihitung selang kepercayaan dari $\hat{p}$ sebagai berikut $$\hat{p}\pm z_{0.025}\sqrt{p_0(1-p_0)/n} \\
= 0.9798 \pm 1.64\sqrt{(0.99) (0.01)/125000}\\
= 0.9798 \pm 4.61 \times 10^{-4}$$

```python
round(norm.ppf(0.05),2)
```

# (Masih) Balada COD di Negara XYZ  
Setelah mengakses data sebelumnya, kini diminta untuk membandingkan apakah terdapat perbedaan persentase keberhasilan COD di Daerah Semangka dan Daerah Pisang. Pada Daerah Semangka, dari 300,000 penjualan dengan sistem COD, hanya 50,000 barang saja yang gagal. Sedangkan pada daerah Pisang, dari 10,000 penjualan dengan sistem COD hanya 50 barang saja yang gagal.

Pembahasan:  
1.  Definisikan terlebih dahulu $H_0$ dan $H_1$ terlebih dahulu
$$
H_0 : p_1 - p_2 = 0 \\
H_1 : p_1 - p_2 \neq 0
$$
2. Hitung $z_{\text{hitung}}$:  
$$
z_{\text{hitung}} = \frac{\hat{p}_1-\hat{p}_2-D_0}{\sqrt{\hat{p}_1(1-\hat{p}_1)/n_1+\hat{p}_2(1-\hat{p}_2)/n_2}}
$$
Dari soal diperoleh bahwa nilai 
$$\hat{p}_1 = \frac{300,000-50,000}{300,000} \approx 0.833\\
\hat{p}_2 = \frac{10,000-50}{10,000} = 0.995$$ 
Sehingga diperoleh $z_{\text{hitung}}$ sebagai berikut
$$
z_{\text{hitung}} = \frac{0.833-0.995-0}{\sqrt{\frac{0.833(1-0.833)}{300,000}+\frac{0.995(1-0.995)}{10,000}}} \approx -164.96
$$
3. Hitung titik kritis  
Perhatikan karna $H_1 : p_1 - p_2 \neq 0$ maka diperoleh bahwa uji ini merupakan uji *two-tailed*. Pada soal, tidak disampaikan secara eksplisit batas signifikansi, $\alpha$, yang akan digunakan. Maka diasumsikan $\alpha = 5\%$. Maka diperoleh titik kritisnya adalah 
$$
z_{0.025} = 1.96
$$
Perhatikan bahwa nilai $z_{\text{hitung}} < -z_{0.025}$ sehingga, $H_0$ **ditolak** atau dengan kata lain, bahwa penjualan dengan sistem COD di daerah Semangka dan Pisang berbeda.

Selang Kepercayaan:
$$
\hat{p}_1- \hat{p}_2 \pm z_{0.025} \text{s.e.}\left( \hat{p}_1- \hat{p}_2 \right)\\
= \hat{p}_1- \hat{p}_2 \pm z_{0.025} \sqrt{\frac{\hat{p}_1(1-\hat{p}_1)}{n_1}+\frac{\hat{p}_2(1-\hat{p}_2)}{n_2}}\\
= (0.833-0.995) \pm 1.96 \sqrt{\frac{0.833(1-0.833)}{300,000}+\frac{0.995(1-0.995)}{10,000}}\\
= -0.162 \pm 1.92 \times 10^{-3}
$$

# Ikan Segar Hasil Dalam Negeri
Setelah bekerja di kantor penjualan barang secara *online*, kini anda direkrut oleh suatu perusahaan penangkapan ikan. Atasan anda sedang kebingungan mengenai hasil tangkapan beberapa bulan yang lalu tidak sesuai dengan harapannya. Dari perhitungan yang dilakukan, diperoleh sampel total berat ikan hasil penangkapan dari perairan Utara tersebut sebesar 8.7 kg dengan banyak ikan sebanyak 10 ekor. Diperoleh pula total sampel berat ikan hasil penangkapan dari perairan Selatan seberat 9 kg dengan banyak ikan 15 ekor. Diketahui pula standar deviasi dari berat ikan sebesar 15 gram dan 25 gram untuk ikan di perairan Utara dan perairan Selatan secara berurutan. Apakah terdapat perbedaan signifikansi antar rataan berat ikan yang ditangkap di Perairan Utara dan di Perairan Selatan ?

Pembahasan:  
1.  Definisikan terlebih dahulu $H_0$ dan $H_1$ terlebih dahulu
$$
H_0 : \mu_1 - \mu_2 = 0 \\
H_1 : \mu_1 - \mu_2 \neq 0
$$
2. Hitung $t_{\text{hitung}}$:
$$
t_{\text{hitung}} = \frac{\bar{X}_1-\bar{X}_2-D_0}{\sqrt{\frac{s^2_1}{n_1}+ \frac{s^2_2}{n_2}}}
$$
Dari soal diperoleh bahwa nilai 
$$\bar{X}_1 = \frac{8,700}{10} = 870 \text{ gram}\\
\bar{X}_2 = \frac{9,000}{15} = 600 \text{ gram}$$ 
Sehingga diperoleh $t_{\text{hitung}}$ sebagai berikut
$$
t_{\text{hitung}} = \frac{870-600-0}{\sqrt{\frac{15^2}{10}+\frac{25^2}{15}}} \approx 33.7061
$$
3. Hitung titik kritis  
Perhatikan karna $H_1 : \mu_1 - \mu_2 \neq 0$ maka diperoleh bahwa uji ini merupakan uji *two-tailed*. Pada soal, tidak disampaikan secara eksplisit batas signifikansi, $\alpha$, yang akan digunakan. Maka diasumsikan $\alpha = 5\%$. Perhatikan bahwa total data sample sebanyak 25 buah, sehingga derajat kebebasan dari $t-statistic$ adalah 23. Maka diperoleh titik kritisnya 
$$
t_{0.025,23} = 2.07031
$$
Perhatikan bahwa nilai $t_{\text{hitung}} > t_{0.025,23}$ sehingga, $H_0$ **ditolak** atau dengan kata lain, bahwa terdapat perbedaan yang signifikan antara penangkapan ikan di perairan Utara dan Selatan.

Selang Kepercayaan:
$$
\bar{X}_1- \bar{X}_2 \pm t_{0.025} \text{s.e.}\left( \bar{X}_1- \bar{X}_2 \right)\\
= \bar{X}_1- \bar{X}_2 \pm t_{0.025} {\sqrt{\frac{s_1^2}{n_1}+\frac{s_2^2}{n_2}}}\\
= (870-600) \pm 2.07031 {\sqrt{\frac{15^2}{10}+\frac{25^2}{15}}}\\
= 270 \pm 16.58 \\
= (253.42, 286.58)
$$
